```python
import pandas as pd
import numpy as np
```

```python
df_matches_goals = pd.read_csv("../data/processed/matches_goals_top50_last3seasons.csv")
df_delay_score = pd.read_csv("../data/processed/top50_players_delay_score.csv")
```

home/away difference

```python
df_matches_goals["nb_home_goals"] = np.where((df_matches_goals["h_a"] == "h") & (df_matches_goals["goals"]>=1), 1, 0)
df_matches_goals["nb_away_goals"] = np.where((df_matches_goals["h_a"] == "a") & (df_matches_goals["goals"]>=1), 1, 0)

df = df_matches_goals.groupby(["player_id", "player"]).agg(
    total_home_goals=("nb_home_goals", "sum"),
    total_away_goals=("nb_away_goals", "sum")
).reset_index()

df["prop_home_goals"] = df["total_home_goals"] / (df["total_home_goals"] + df["total_away_goals"])
df["prop_away_goals"] = df["total_away_goals"] / (df["total_home_goals"] + df["total_away_goals"])
```

```python
big_diff = df[(df["prop_home_goals"] > 0.65) | (df["prop_away_goals"] > 0.65)]
big_diff.sort_values(by="prop_home_goals", ascending=False)
```
