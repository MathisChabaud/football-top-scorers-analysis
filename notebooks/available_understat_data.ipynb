```python
from understatapi import UnderstatClient
import pandas as pd

understat = UnderstatClient()
```

Function to explore the structure of the retrieved understat data.

```python
def explore_structure(data, indent=0, max_depth=3, max_items=2):

    prefix = "  " * indent
    data_type = type(data).__name__
    
    if isinstance(data, dict):
        print(f"{prefix} Dictionary ({data_type}) - {len(data)} keys")
        for i, (key, value) in enumerate(data.items()):
            if i >= max_items:
                print(f"{prefix}  ... and {len(data) - max_items} other keys")
                break
            print(f"{prefix} '{key}': ", end="")
            if max_depth > 1:
                print()
                explore_structure(value, indent + 2, max_depth - 1, max_items)
            else:
                print(type(value).__name__, end="")
                if isinstance(value, (list, dict)):
                    print(f" ({len(value)} elements)")
                else:
                    print(f" = {repr(value)[:50]}")
                    
    elif isinstance(data, list):
        print(f"{prefix} List ({data_type}) - {len(data)} elements")
        for i, item in enumerate(data[:max_items]):
            print(f"{prefix}  [{i}]: ", end="")
            if max_depth > 1:
                print()
                explore_structure(item, indent + 2, max_depth - 1, max_items)
            else:
                print(type(item).__name__)
        if len(data) > max_items:
            print(f"{prefix}  ... and {len(data) - max_items} other elements")
            
    else:
        print(f"{prefix} {data_type} = {repr(data)[:100]}")
```
